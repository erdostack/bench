# Lineage tour

A quick look at which published datasets are built from which.

```python
import ladybug as lb

db = lb.Database("../var/lineage.lbdb")
conn = lb.Connection(db)
conn.execute("MATCH (d:Dataset)-[:DERIVED_FROM*1..3]->(s:Dataset) RETURN d.slug, s.slug LIMIT 20").get_as_df()
```
