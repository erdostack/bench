# Who connects the chapters?

Betweenness on the weighted graph (weights inverted to distances).

```python
import sys
sys.path.insert(0, '..')
from lesmis.load import graph
g = graph()
dist = [1 / w for w in g.es['weight']]
bt = g.betweenness(weights=dist)
sorted(zip(bt, g.vs['name']), reverse=True)[:5]
```
