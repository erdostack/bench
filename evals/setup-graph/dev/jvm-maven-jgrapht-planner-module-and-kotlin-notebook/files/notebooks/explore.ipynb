# Exploring the hut graph
Scratch notebook from the first planning season.

```kotlin
@file:DependsOn("org.jgrapht:jgrapht-core:1.3.0")
import org.jgrapht.graph.*
import org.jgrapht.alg.shortestpath.*
```

```kotlin
val g = SimpleDirectedWeightedGraph<String, DefaultWeightedEdge>(DefaultWeightedEdge::class.java)
java.io.File("../data/trails.csv").readLines().drop(1).forEach { line ->
    val (a, b, h) = line.split(",")
    g.addVertex(a); g.addVertex(b)
    g.setEdgeWeight(g.addEdge(a, b), h.toDouble())
}
DijkstraShortestPath(g).getPath("H07", "H09")
```
