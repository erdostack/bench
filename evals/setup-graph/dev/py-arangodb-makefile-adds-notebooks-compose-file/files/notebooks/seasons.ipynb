# Visits per season

```python
import os
from arango import ArangoClient
db = ArangoClient(hosts=os.environ['ARANGO_URL']).db('pollen', username='root', password=os.environ.get('ARANGO_ROOT_PASSWORD', ''))
list(db.aql.execute('FOR e IN visited COLLECT s = e.season WITH COUNT INTO n RETURN {s, n}'))
```
